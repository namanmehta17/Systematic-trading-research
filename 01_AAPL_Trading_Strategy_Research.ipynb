## Execution and evaluation design
Signals observed at close execute at the **next session's open**, never the same close. Adjustment factor = adjusted close / raw close; apply it to raw open too. These are adjusted-price return units, not literal historical share counts. Dividend reinvestment is approximated by the vendor adjustment factors. No shorting, leverage, taxes or cash interest; 10 bps per traded side approximates fees and slippage, not an observed brokerage quote.

First 252 observations are the initial selection window. Monthly re-selection uses only trailing returns then known. Every evaluated account starts flat on the same date. Fixed strategies were chosen before inspecting results in this notebook. Historical adjusted data are retrospective, not a point-in-time feed; the short history, single selected surviving stock and unequal indicator warm-up reduce robustness.

```python
"""Long-only, next-open, self-financing research backtester. No live orders."""
import numpy as np
import pandas as pd
from scipy.optimize import brentq


def prepare_prices(frame):
    """Require a common complete calendar; never silently forward-fill prices."""
    required = {'date', 'symbol', 'open', 'close'}
    if not required <= set(frame):
        raise ValueError(f'Required columns: {sorted(required)}')
    f = frame.copy(); f['date'] = pd.to_datetime(f['date'], errors='raise')
    if f.empty or f[list(required)].isna().any().any() or f.duplicated(['date', 'symbol']).any():
        raise ValueError('Missing data, empty input or duplicate date-symbol pairs')
    if not np.isfinite(f[['open', 'close']].to_numpy(dtype=float)).all() or (f[['open', 'close']] <= 0).any().any():
        raise ValueError('Prices must be finite and positive')
    op = f.pivot(index='date', columns='symbol', values='open').sort_index()
    cl = f.pivot(index='date', columns='symbol', values='close').reindex_like(op)
    if op.isna().any().any() or cl.isna().any().any():
        raise ValueError('Use a complete common trading calendar; document exclusions')
    return op, cl


def wilder_rsi(series, window=14):
    """Wilder smoothing seeded with the first window of changes."""
    x = series.to_numpy(float); out = np.full(len(x), np.nan)
    if len(x) <= window:
        return pd.Series(out, index=series.index)
    delta = np.diff(x); gains = np.maximum(delta, 0); losses = np.maximum(-delta, 0)
    g, l = gains[:window].mean(), losses[:window].mean()
    for i in range(window, len(x)):
        if i > window:
            g = ((window-1)*g + gains[i-1])/window
            l = ((window-1)*l + losses[i-1])/window
        out[i] = 50 if g == l == 0 else (100 if l == 0 else 100 - 100/(1+g/l))
    return pd.Series(out, index=series.index)


def held_position(entry, exit_):
    """Retain a position between signals; a sell signal means flat, not short."""
    held = False; result = []
    for buy, sell in zip(entry, exit_):
        if held and sell: held = False
        elif not held and buy: held = True
        result.append(float(held))
    return np.array(result)


def signals(close):
    result = {key: pd.DataFrame(0.0, index=close.index, columns=close.columns)
              for key in ['SMA_50_200', 'RSI_30_70', 'Bollinger_20_2']}
    for s in close:
        x = close[s]; slow = x.rolling(200).mean(); fast = x.rolling(50).mean()
        result['SMA_50_200'][s] = ((fast > slow) & slow.notna()).astype(float)
        rsi = wilder_rsi(x)
        result['RSI_30_70'][s] = held_position(rsi < 30, rsi > 70)
        mid = x.rolling(20).mean(); sd = x.rolling(20).std(ddof=0)
        result['Bollinger_20_2'][s] = held_position(x < mid-2*sd, x > mid+2*sd)
    # Each asset has a fixed maximum sleeve of 1 / number of assets.
    return {k: v/len(close.columns) for k, v in result.items()}


def simulate(op, close, close_targets, cost_bps=10, initial=100000.0):
    """Close-t target weights execute at open t+1; open-to-close marked daily.

    Target weights are reset daily, including within an active position.
    Costs apply to absolute traded notional on buys and sells. Cash earns zero.
    Post-cost sizing is solved so the account never borrows to pay costs.
    """
    if not op.index.equals(close.index) or not op.columns.equals(close.columns):
        raise ValueError('Open and close matrices must align')
    if not close_targets.index.equals(close.index) or not close_targets.columns.equals(close.columns):
        raise ValueError('Targets must align exactly with prices')
    w = close_targets.to_numpy(float)
    if not np.isfinite(w).all() or (w < 0).any() or (w.sum(axis=1) > 1+1e-10).any():
        raise ValueError('Weights must be finite, nonnegative and sum to at most one')
    if not 0 <= cost_bps < 1000 or initial <= 0:
        raise ValueError('Invalid transaction costs or initial capital')
    if (op <= 0).any().any() or (close <= 0).any().any() or not np.isfinite(op).all().all() or not np.isfinite(close).all().all():
        raise ValueError('Invalid prices')
    cost = cost_bps/10000; shares = np.zeros(op.shape[1]); cash = initial
    rows, trades = [], []
    for i, date in enumerate(op.index):
        prices = op.iloc[i].to_numpy(float); current = shares*prices
        pre = cash + current.sum(); target = w[i-1] if i else np.zeros_like(shares)
        def balance(post): return post + cost*np.abs(target*post-current).sum() - pre
        post = pre if cost == 0 else brentq(balance, 0, pre)
        dollars = target*post-current; fee = cost*np.abs(dollars).sum()
        cash -= dollars.sum() + fee
        shares += dollars/prices
        if cash < -1e-6: raise AssertionError('Unexpected borrowing')
        nav = cash + np.dot(shares, close.iloc[i].to_numpy(float))
        for j, amount in enumerate(dollars):
            if abs(amount) > 1e-6:
                trades.append((date, op.columns[j], amount, prices[j], abs(amount)*cost))
        rows.append((date, nav, cash, fee, np.abs(dollars).sum()/pre))
    daily = pd.DataFrame(rows, columns=['date', 'nav', 'cash', 'cost', 'turnover']).set_index('date')
    daily['return'] = daily.nav.pct_change().fillna(daily.nav.iloc[0]/initial-1)
    daily['drawdown'] = daily.nav / daily.nav.cummax().clip(lower=initial)-1
    ledger = pd.DataFrame(trades, columns=['date', 'symbol', 'notional', 'execution_price', 'cost'])
    return daily, ledger


def walk_forward_targets(close, candidates, candidate_runs, train=504):
    """Select monthly using ONLY trailing net returns known at the signal close.

    The first signal after train observations executes on the next session.
    Negative best trailing Sharpe -> cash. This is a research rule, not a claim
    of optimal model selection. A chosen model may already have an open signal.
    """
    target = pd.DataFrame(0.0, index=close.index, columns=close.columns)
    decisions = []; chosen = None; last_month = None
    for i in range(train-1, len(close)):
        date = close.index[i]; month = date.to_period('M')
        if month != last_month:
            scores = {}
            for name, daily in candidate_runs.items():
                r = daily['return'].iloc[i-train+1:i+1]
                sd = r.std(ddof=1)
                scores[name] = np.sqrt(252)*r.mean()/sd if sd > 1e-12 else -np.inf
            best = max(scores, key=scores.get)
            chosen = best if scores[best] > 0 else None
            decisions.append((date, chosen or 'CASH', scores[best], close.index[i-train+1]))
            last_month = month
        if chosen: target.iloc[i] = candidates[chosen].iloc[i]
    return target, pd.DataFrame(decisions, columns=['signal_date', 'strategy', 'training_sharpe', 'training_start'])


def metrics(daily):
    r = daily['return']; wealth = (1+r).cumprod(); peak = wealth.cummax().clip(lower=1)
    sd = r.std(ddof=1)
    return pd.Series({'total_return': wealth.iloc[-1]-1,
                      'CAGR_252_sessions': wealth.iloc[-1]**(252/len(r))-1,
                      'volatility': sd*np.sqrt(252),
                      'Sharpe_zero_cash_rate': r.mean()/sd*np.sqrt(252) if sd > 1e-12 else np.nan,
                      'max_drawdown': (wealth/peak-1).min(),
                      'annual_one_way_turnover': daily.turnover.mean()*252,
                      'cost_dollars': daily.cost.sum()})
```

```python
factor = raw['AAPL.Adjusted']/raw['AAPL.Close']
frame = pd.DataFrame({'date':raw.index, 'symbol':'AAPL', 'open':raw['AAPL.Open'].to_numpy()*factor.to_numpy(), 'close':raw['AAPL.Adjusted'].to_numpy()})
op, close = prepare_prices(frame)
candidates = signals(close)

def research(cost_bps=10):
    full_runs = {k: simulate(op,close,v,cost_bps)[0] for k,v in candidates.items()}
    target, decisions = walk_forward_targets(close,candidates,full_runs,train=252)
    targets = {**candidates,'Monthly_selector':target,'Buy_and_hold':pd.DataFrame(1.,index=close.index,columns=close.columns)}
    runs, ledgers = {}, {}
    for k,v in targets.items():
        runs[k],ledgers[k] = simulate(op.iloc[251:],close.iloc[251:],v.iloc[251:],cost_bps)
    summary = pd.DataFrame({k:metrics(v.iloc[1:]) for k,v in runs.items()}).T
    return summary,runs,ledgers,decisions
summary,runs,ledgers,decisions = research()
display(summary.round(4)); display(decisions)
summary.to_csv(OUT/'aapl_strategy_metrics.csv')
ledgers['Monthly_selector'].to_csv(OUT/'aapl_trade_ledger.csv',index=False)
```

## Validation: causality and accounting
Changing future prices must not change earlier signals. Also check no immediate execution at the signal close, nonnegative cash, and consistent cost accounting.

```python
cut=350
changed=close.copy();changed.iloc[cut:]*=1.5
for key,val in signals(changed).items():
    pd.testing.assert_frame_equal(val.iloc[:cut],candidates[key].iloc[:cut])
tiny_op=op.iloc[:4]; tiny_cl=close.iloc[:4]
target=pd.DataFrame(1.,index=tiny_cl.index,columns=tiny_cl.columns)
d,l=simulate(tiny_op,tiny_cl,target)
assert l.date.min()==tiny_cl.index[1] and d['cash'].min()>-1e-6
assert abs(d.cost.sum()-l.cost.sum())<1e-7
for d in runs.values(): assert d.cash.min()>-1e-6
print('PASS: causal signals, next-open execution, cash constraint and fee reconciliation')
b=summary.loc['Buy_and_hold'];w=summary.loc['Monthly_selector']
print(f"Selector net return {w.total_return:.2%}; buy-and-hold {b.total_return:.2%}; gap {w.total_return-b.total_return:+.2%}.")
print(f"Selector max drawdown {w.max_drawdown:.2%}; benchmark {b.max_drawdown:.2%}.")
print('These results describe this frozen period only; they do not establish persistent alpha.')
```

## Investment interpretation and next research steps
Read the net-return gap alongside drawdown and time in cash. A lower drawdown may simply reflect lower equity exposure. Cost sensitivity reveals whether frequent switching consumes the edge. A negative result is useful evidence, not a reason to change parameters until the chart looks attractive.

Before extending a trading claim: expand to an ex-ante universe including delisted securities, use longer history and multiple market regimes, compare with cash-interest and matched-exposure benchmarks, obtain timestamped executable quotes, and preserve a final untouched test set. Do not describe this limited study as a validated trading strategy.

Cost sensitivity re-runs monthly selection using net training returns at each cost level. Therefore the selector can choose different strategies, and its realized return need not decrease monotonically with costs. Fixed-rule comparisons isolate costs more directly.
